```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Markdown
repo = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "reported_runs/analysis/packaged.py").is_file())
sys.path.insert(0, str(repo)) if str(repo) not in sys.path else None
from reported_runs.analysis.presentation import load_results, NotebookReport, label_conditions
from reported_runs.analysis.packaged import derived_frame, bundle_provenance
from reported_runs.analysis.plotting import trajectory_grid, effect_grid, rank_grid, association_grid
%matplotlib inline
results = load_results()
report = NotebookReport(results, 'rq3_profile_review')
sources = results["sources"]
trajectory = label_conditions(results["trajectory"])
summaries = results["summaries"]
effect_summaries = results["effect_summaries"]
scales = report.scales
```

## Pinned sample and selection


```python
assessments = derived_frame('profile_assessments')
selection = derived_frame('profile_review_selection')
report.table(selection, 'rq3_profile_review_sampling', "", columns=['seed', 'population_seed', 'eligible_users', 'requested_sample', 'selected_users'])
report.table(assessments, 'rq3_profile_review_case_index', "", columns=['case_id', 'seed', 'population_seed', 'review_order', 'applied_change_count', 'first_change_iteration', 'last_change_iteration'])
```

## Recount recorded human assessments


```python
from reported_runs.analysis.packaged import profile_frequencies
from reported_runs.analysis.profile_review_views import frequency_table, CATEGORY_COLUMNS, STATUS_COLUMNS
frequencies = profile_frequencies(assessments)
original_frequencies = derived_frame('profile_review_frequencies_original')
pd.testing.assert_frame_equal(frequencies.astype({'seed': 'Float64'}).reset_index(drop=True), original_frequencies.astype({'seed': 'Float64'}).reset_index(drop=True), check_dtype=False)
for dimension, name, caption, columns in [('review_status', 'completion', "", STATUS_COLUMNS), ('extent_of_change', 'extent', "", CATEGORY_COLUMNS), ('support_in_available_evidence', 'support', "", CATEGORY_COLUMNS)]:
    report.table(frequency_table(frequencies, dimension), 'rq3_profile_review_' + name, "", column_format=columns, source_frame=assessments)
```
